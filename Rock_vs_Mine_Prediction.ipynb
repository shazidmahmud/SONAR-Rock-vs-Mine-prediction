```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
data=pd.read_csv("Copy of sonar data.csv",header=None)
```

```python
print(data)
```

```python
data.head()
```

```python
data[60].value_counts()
```

```python
data.groupby(60).mean()
```

```python
#separating data and labels
X=data.drop(columns=60,axis =1)
Y=data[60];
```

```python
print(X)
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.1,stratify=Y,random_state=1)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
model=LogisticRegression()
```

```python
print(X_train)
print(Y_train)
```

```python
model.fit(X_train, Y_train)
```

```python
#accuracy of the training data
XtrainPredict=model.predict(X_train)
acc=accuracy_score(XtrainPredict, Y_train)
print("ACCURACYyyyyyyy: ",acc)
```

```python
XtestPredict=model.predict(X_test)
acc2=accuracy_score(XtestPredict, Y_test)
print("FINAL : ",acc2)
```
